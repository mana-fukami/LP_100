# 問2.6

```python
class DataManager:
  def __init__(self,x,y,z):
    self.x=x
    self.y=y
    self.z=z

  def add_x(self,delta):
    self.x+=delta

  def add_y(self,delta):
    self.y+=delta

  def add_z(self,delta):
    self.x+=delta

  def sum(self):
    return self.x+self.y+self.z
```

```python
data_manager = DataManager(2, 3, 5)
print(data_manager.sum())  # => 10
data_manager.add_x(4)      # => data_manager.x の値が 2 から 6 に更新される
print(data_manager.sum())  # => 14
data_manager.add_y(0)      # => data_manager.y の値が 3 から 3 に更新される
print(data_manager.sum())  # => 14
data_manager.add_z(-9)     # => data_manager.z の値が 5 から -4 に更新される
print(data_manager.sum())  # => 5
```

# 問2.7

```python
def f(a):
  a=[6,7,8]

def g(a):
  a.append(1)
```

```python
def somefunction():
  a0=[1,2,3]
  f(a0)
  print(a0)

  a1=[1,2,3]
  g(a1)
  print(a1)

"""
結果↓
[1,2,3]
[1,2,3,1]
"""
somefunction()
```

説明：
f(a)では、aに代入を行っているが、グローバル変数として扱っていないので、戻り値として返さなければ書き換わることはない。
一方g(a)では、a.append(1)自体が、引数として渡ったリストの末尾に1を追加し中身を上書きするという操作であるから、返り値を返さなくても書き換わる。

```python
def vector_sum(x,y):
  sum=[]
  for i in range(len(x)):
    sum.append(x[i]+y[i])
  return sum

x=[1,2,3]
y=[8,1,2]
answer=vector_sum(x,y)
print(answer)
```

2.

```python
def matrix_sum(X,Y):
  sum=[]
  for i in range(len(X)):
    sum.append(vector_sum(X[i],Y[i]))
  return sum

X=[[1,2,3],
   [4,5,6]]
Y=[[8,1,2],
   [-1,0,-2]]
answer=matrix_sum(X,Y)
print(answer)
```

3.

```python
def matrix_vector_product(X,y):
  answer=[]
  for i in range(len(X)):
    product=0
    for j in range(len(X[i])):
      product+=X[i][j]*y[j]
    answer.append(product)
  return answer

X=[[1,2,3],
   [4,5,6]]
y=[8,1,2]
answer=matrix_vector_product(X,y)
print(answer)
```

4.

```python
def matrix_product(X,Y):
  answer=[]
  for x in X:
    row=[]
    for i in range(len(X)):
      product=0
      for j in range(len(Y)):
        product+=x[j]*Y[j][i]
      row.append(product)
    answer.append(row)
  return answer

X=[[1,2,3],
   [4,5,6]]
Y=[[8,1],
   [-1,0],
   [0,1]]
answer=matrix_product(X,Y)
print(answer)
```

```python
def matrix_transpose(X):
  answer=[]
  for i in range(len(X[0])):
    row=[]
    for j in range(len(X)):
      row.append(X[j][i])
    answer.append(row)
  return answer

X = [[1, 2, 3],
     [4, 5, 6]]
answer = matrix_transpose(X)
print(answer)
```

# 問6.5

```python
import math
import random

def log_like(theta):
  logL=0
  for i in range(0,30,1):
    l=math.log(theta[0])
    logL+=l
  for i in range(1,6,1):
    for j in range(6):
      logL+=math.log(theta[i])
  return logL

def new_theta():
  theta_sum=1
  new_theta=[]
  for i in range(6):
    new_theta.append(random.uniform(0,theta_sum))
    theta_sum-=new_theta[i]
  return new_theta

theta=[1/6,1/6,1/6,1/6,1/6,1/6]
for i in range(5):
  formatted=["{:.4f}".format(t) for t in theta]
  print(formatted)
  print(log_like(theta))
  theta=new_theta()
```
